```python
import re
```

```python
text = 'I love machine learning'
```

```python
pattern = 'machine'
```

```python
print(re.search(pattern,text))
```

```python
data = 'machine learning'
```

```python
print(re.match('^machine',data))  # start with machine # we can use search also
```

```python
print(re.search('learning$', data)) # end with learning
```

```python
# digit --> \d
# word --> \w
# space --> \s
```

```python
d = 'My number is 2345654'
```

```python
re.findall(r'\d',d)
```

```python
# [cb]at
```

```python
text = 'cat bat rat mat'
re.findall(r'[cb]at',text)
```

```python
re.findall(r'[cbr]at',text)
```

```python
d = 'aaa aa a'
re.findall(r'a+',d)
```

```python
text = 'Contact:; hr@codegemini.com, python@codegemini.com, xyz.company'

re.findall(r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-z]{2,}", text)
```

```python
# a-z lowercase
# A-Z uppercase
# 0-9 digits
# _%+- allowed symbols
# [] any of this chars
# + one or more
# ? optional
```

```python
d = 'My number is 1234567890'
```

```python
re.findall(r'\d{10}',d)
```

```python
raw_text = 'Visit https://google.com'
```

```python
re.findall(r'https?://\S+',raw_text)
```
