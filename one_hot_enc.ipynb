```python
# Lowercase + stopwords
from nltk.corpus import stopwords
stop_words = stopwords.words('english')

processed_tokens = []
for sentence in corpus:
    words = word_tokenize(sentence.lower())
    filtered_words = [w for w in words if w not in stop_words]
    processed_tokens.append(" ".join(filtered_words))
```

```python
for w in words:
    if w not in stop_words:
        print(w)
```

# One-hot-encoding

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_corpus = []

for sentence in processed_tokens:
    words = sentence.split()
    lem_words = [lemmatizer.lemmatize(w) for w in words]
    lemmatized_corpus.append(' '.join(lem_words))
print(lemmatized_corpus)
```

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```python
words = list(set(' '.join(lemmatized_corpus).split()))
print(words)
```

```python
words = np.array(words).reshape(-1,1)
```

```python
words
```

```python
encoder = OneHotEncoder(sparse_output=False)
one_hot = encoder.fit_transform(words)
```
